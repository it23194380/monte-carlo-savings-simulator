# Monte Carlo Simulation of Savings Growth Under Uncertain Investment Returns and Inflation

## 1. Title and Summary

This project investigates the probability of achieving a long-term savings target under uncertain investment returns and inflation.

A Monte Carlo simulation generates 10,000 possible savings outcomes over 10 years, assuming monthly contributions of LKR 15,000 and an inflation-adjusted target of LKR 2,000,000.

The simulation evaluates the probability of reaching the target, examines uncertainty in final savings, and compares different financial planning scenarios.

## 2. Problem and Research Question

### Research Question

What is the probability that an individual saving LKR 15,000 per month will accumulate the equivalent of LKR 2,000,000 in today's purchasing power within 10 years, considering uncertain investment returns and inflation?

### Objectives

1. Model monthly savings accumulation.
2. Simulate uncertain investment returns and inflation.
3. Estimate the probability of achieving the savings target.
4. Verify the simulation against a deterministic savings formula.
5. Investigate how changes in contributions, returns, and inflation affect the probability of success.

## 3. Model, Assumptions and Equations

### 3.1 Model Overview

This project uses Monte Carlo simulation to estimate the probability of achieving a long-term savings target under uncertain investment returns and inflation.

An individual contributes LKR 15,000 at the end of every month for 10 years. Investment returns and inflation vary randomly each month.

The model generates 10,000 independent simulation paths, each representing a possible financial outcome over 120 months.

The final nominal savings are adjusted for cumulative inflation and compared against a target of LKR 2,000,000 expressed in today's purchasing power.

### 3.2 Baseline Parameters

| Parameter | Value |
|---|---|
| Monthly contribution | LKR 15,000 |
| Initial savings | LKR 0 |
| Savings duration | 10 years (120 months) |
| Real savings target | LKR 2,000,000 |
| Mean annual investment return | 8% |
| Annual investment return volatility | 12% |
| Mean annual inflation | 5% |
| Annual inflation volatility | 2% |
| Monte Carlo simulations | 10,000 |
| Random seed | 42 |

The financial parameters are illustrative modelling assumptions rather than estimates calibrated from historical Sri Lankan financial data.

### 3.3 Monthly Savings Accumulation

Savings grow through investment returns and regular monthly contributions.

The savings balance is updated using:

$$S_{t+1}=S_t(1+r_t)+C$$

Where:

- $S_t$ = nominal savings balance at month $t$
- $r_t$ = investment return during month $t$
- $C$ = fixed monthly contribution (LKR 15,000)
- $S_0=0$ = initial savings balance

Contributions are deposited at the end of each month, after applying that month's investment return.

### 3.4 Stochastic Investment Returns

Investment returns are uncertain and are modelled using a normal distribution.

$$r_t \sim N\left(\frac{\mu_r}{12},\frac{\sigma_r^2}{12}\right)$$

Where:

- $\mu_r=0.08$ = assumed annualized mean investment return
- $\sigma_r=0.12$ = assumed annualized return volatility

The monthly mean and standard deviation are:

$$\mu_{r,m}=\frac{0.08}{12}$$

$$\sigma_{r,m}=\frac{0.12}{\sqrt{12}}$$

The simulation generates a different random investment return for each month and simulation path.

The normal distribution is a simplified assumption and may not fully represent extreme financial market movements.

### 3.5 Stochastic Inflation

Inflation reduces the purchasing power of accumulated savings.

Monthly inflation is also modelled using a normal distribution:

$$\pi_t \sim N\left(\frac{\mu_\pi}{12},\frac{\sigma_\pi^2}{12}\right)$$

Where:

- $\mu_\pi=0.05$ = assumed annualized mean inflation
- $\sigma_\pi=0.02$ = assumed annualized inflation volatility

Monthly inflation parameters are:

$$\mu_{\pi,m}=\frac{0.05}{12}$$

$$\sigma_{\pi,m}=\frac{0.02}{\sqrt{12}}$$

The cumulative inflation index is updated using:

$$I_{t+1}=I_t(1+\pi_t)$$

Where:

- $I_t$ = cumulative inflation index at month $t$
- $\pi_t$ = monthly inflation rate
- $I_0=1$ = initial inflation index

### 3.6 Inflation-Adjusted Savings

Nominal savings are converted into real savings using the cumulative inflation index.

$$S_{\mathrm{real},t}=\frac{S_t}{I_t}$$

Where:

- $S_t$ = nominal savings
- $I_t$ = cumulative inflation index
- $S_{\mathrm{real},t}$ = savings measured in today's purchasing power

The savings target is considered achieved if:

$$S_{\mathrm{real},120}\geq 2,000,000$$

### 3.7 Monte Carlo Probability Estimation

The simulation generates 10,000 possible financial outcomes.

For each simulation, the final real savings are compared against the target.

The estimated probability of success is:

$$\hat{p}=\frac{1}{N}\sum_{i=1}^{N}X_i$$

Where:

- $N$ = number of simulation paths
- $X_i=1$ if simulation $i$ achieves the target
- $X_i=0$ otherwise
- $\hat{p}$ = estimated probability of success

A 95% Wilson confidence interval is calculated to quantify Monte Carlo sampling uncertainty in the estimated probability.

### 3.8 Model Assumptions

The baseline model makes the following assumptions:

1. Monthly contributions remain fixed at LKR 15,000.
2. Contributions are made at the end of each month.
3. Initial savings are zero.
4. Monthly investment returns follow a normal distribution.
5. Monthly inflation rates follow a normal distribution.
6. Investment returns and inflation shocks are independent of each other and across months.
7. The same distribution parameters apply throughout the 10-year period.
8. Investment fees, taxes, and transaction costs are excluded.
9. Contributions are nominal amounts and are not automatically increased with inflation.
10. The model uses discrete monthly compounding and inflation adjustment.

These assumptions simplify the financial system and make the simulation reproducible.

### 3.9 Connection to Modelling and Simulation Methods

The project demonstrates three modelling and simulation topics.

**1. Monte Carlo Simulation**

Random investment returns and inflation rates are generated across 10,000 simulation paths to estimate the probability of reaching the savings target.

**2. Random Variate Generation and Statistical Modelling**

Normally distributed random variables are generated using NumPy. Distribution parameters represent the assumed uncertainty in financial returns and inflation.

**3. Numerical Integration of Ordinary Differential Equations**

A supplementary deterministic continuous-time savings model is considered:

$$\frac{dS}{dt}=rS+c$$

Forward Euler's method approximates this equation:

$$S_{k+1}=S_k+\Delta t(rS_k+c)$$

The Euler approximation is compared against the exact continuous-time solution to demonstrate numerical convergence.

This ODE experiment is supplementary to the main discrete-time Monte Carlo simulation.

### 3.10 Reproducibility

The simulation uses NumPy's random number generator with a fixed seed of 42.

Using the same seed, parameters, software environment, and execution order allows the random scenarios and results to be reproduced.

The model uses vectorized NumPy arrays for efficient simulation of 10,000 financial paths over 120 months.

### Investment Returns and Inflation

Investment returns and inflation are uncertain, so the model generates random values for each month.

We assume annual investment returns and inflation follow normal distributions, with parameters converted to monthly values.

Monthly investment return:

$$r_t \sim N\left(\frac{\mu_r}{12}, \frac{\sigma_r^2}{12}\right)$$

Monthly inflation rate:

$$\pi_t \sim N\left(\frac{\mu_\pi}{12}, \frac{\sigma_\pi^2}{12}\right)$$

Where:

- $\mu_r$ = assumed annual mean investment return (8%)
- $\sigma_r$ = assumed annual return volatility (12%)
- $\mu_\pi$ = assumed annual mean inflation (5%)
- $\sigma_\pi$ = assumed annual inflation volatility (2%)

The monthly standard deviations are calculated by dividing annual volatility by $\sqrt{12}$.

The initial model assumes independent monthly returns and inflation shocks. Correlation can be introduced as a later extension.

The cumulative inflation index is updated using:

$$I_{t+1}=I_t(1+\pi_t)$$

Inflation-adjusted savings are:

$$S_{\mathrm{real},t}=\frac{S_t}{I_t}$$

## 4. Implementation

```python
import numpy as np
import matplotlib.pyplot as plt

# Simulation parameters
MONTHLY_CONTRIBUTION = 15000
YEARS = 10
MONTHS = YEARS * 12
TARGET = 2_000_000
N_SIMULATIONS = 10_000
SEED = 42

# Initial savings and inflation index
initial_savings = 0.0
initial_inflation_index = 1.0

rng = np.random.default_rng(SEED)

print("Monte Carlo Savings Simulation")
print(f"Monthly contribution: LKR {MONTHLY_CONTRIBUTION:,}")
print(f"Duration: {YEARS} years")
print(f"Target (today's LKR): {TARGET:,}")
print(f"Simulation paths: {N_SIMULATIONS:,}")
```

```python
# Assumed annual parameters
ANNUAL_RETURN_MEAN = 0.08
ANNUAL_RETURN_VOLATILITY = 0.12

ANNUAL_INFLATION_MEAN = 0.05
ANNUAL_INFLATION_VOLATILITY = 0.02

# Convert annual parameters to monthly
monthly_return_mean = ANNUAL_RETURN_MEAN / 12
monthly_return_std = ANNUAL_RETURN_VOLATILITY / np.sqrt(12)

monthly_inflation_mean = ANNUAL_INFLATION_MEAN / 12
monthly_inflation_std = ANNUAL_INFLATION_VOLATILITY / np.sqrt(12)

# Generate random monthly returns and inflation rates
monthly_returns = rng.normal(
    monthly_return_mean,
    monthly_return_std,
    size=(N_SIMULATIONS, MONTHS)
)

monthly_inflation = rng.normal(
    monthly_inflation_mean,
    monthly_inflation_std,
    size=(N_SIMULATIONS, MONTHS)
)

print("Random financial scenarios generated successfully!")
print("Returns shape:", monthly_returns.shape)
print("Inflation shape:", monthly_inflation.shape)
```

### Monthly Savings Simulation

The simulation calculates savings accumulation over 120 months for 10,000 possible financial scenarios.

At each month:

1. The existing savings balance earns a randomly generated investment return.
2. A fixed monthly contribution is added.
3. The cumulative inflation index is updated.
4. The inflation-adjusted savings balance is calculated.

The model uses the following equations:

$$S_{t+1}=S_t(1+r_t)+C$$

$$I_{t+1}=I_t(1+\pi_t)$$

$$S_{\mathrm{real},t}=\frac{S_t}{I_t}$$

A savings goal is considered achieved when the final inflation-adjusted savings balance is greater than or equal to the target.

The probability of success is estimated as:

$$\hat{P}(\mathrm{success})=\frac{\text{Number of successful simulations}}{\text{Total simulations}}$$

```python
# Arrays to store savings and inflation over time
savings_paths = np.zeros((N_SIMULATIONS, MONTHS + 1))
inflation_paths = np.ones((N_SIMULATIONS, MONTHS + 1))

# Simulate savings growth month by month
for month in range(MONTHS):

    # Apply monthly investment return and add contribution
    savings_paths[:, month + 1] = (
        savings_paths[:, month] * (1 + monthly_returns[:, month])
        + MONTHLY_CONTRIBUTION
    )

    # Update cumulative inflation
    inflation_paths[:, month + 1] = (
        inflation_paths[:, month] * (1 + monthly_inflation[:, month])
    )

# Calculate inflation-adjusted savings
real_savings_paths = savings_paths / inflation_paths

# Final savings values
final_nominal_savings = savings_paths[:, -1]
final_real_savings = real_savings_paths[:, -1]

# Determine which simulations achieved the target
successful_simulations = final_real_savings >= TARGET

# Calculate probability of reaching the target
success_probability = np.mean(successful_simulations)

# Calculate summary statistics
mean_final_savings = np.mean(final_real_savings)
median_final_savings = np.median(final_real_savings)
std_final_savings = np.std(final_real_savings, ddof=1)

print("MONTE CARLO SAVINGS SIMULATION RESULTS")
print("-------------------------------------")
print(f"Number of simulations: {N_SIMULATIONS:,}")
print(f"Monthly contribution: LKR {MONTHLY_CONTRIBUTION:,}")
print(f"Duration: {YEARS} years")
print(f"Target (today's value): LKR {TARGET:,.2f}")
print()
print(f"Mean final real savings: LKR {mean_final_savings:,.2f}")
print(f"Median final real savings: LKR {median_final_savings:,.2f}")
print(f"Standard deviation: LKR {std_final_savings:,.2f}")
print(f"Probability of reaching target: {success_probability:.2%}")


# Basic consistency checks
assert savings_paths.shape == (N_SIMULATIONS, MONTHS + 1)
assert inflation_paths.shape == (N_SIMULATIONS, MONTHS + 1)
assert np.all(np.isfinite(real_savings_paths))
assert np.all(inflation_paths > 0)
assert 0 <= success_probability <= 1

print("\nBasic consistency checks passed!")
```

## 5. Verification

### Verification Against the Analytical Future-Value Formula

To verify the savings accumulation algorithm, we compare its output against the analytical future-value formula for regular monthly contributions.

For a constant monthly investment return, the future value of end-of-month contributions is:

$$FV=C\frac{(1+r)^n-1}{r}$$

Where:

- $FV$ = final accumulated savings (LKR)
- $C$ = monthly contribution (LKR)
- $r$ = constant monthly investment return
- $n$ = number of months

For $r=0$, the limiting case is:

$$FV=Cn$$

The simulation is verified by setting investment-return volatility and inflation to zero, then comparing the simulated nominal savings with the analytical result.

A small numerical tolerance is used to account for floating-point rounding.

```python
# Verification: constant monthly return, no inflation

test_monthly_return = 0.08 / 12
test_contribution = MONTHLY_CONTRIBUTION
test_months = MONTHS

# Analytical future value
analytical_fv = test_contribution * (
    (1 + test_monthly_return) ** test_months - 1
) / test_monthly_return

# Simulated future value using the same update rule
simulated_fv = 0.0

for month in range(test_months):
    simulated_fv = (
        simulated_fv * (1 + test_monthly_return)
        + test_contribution
    )

# Calculate numerical error
absolute_error = abs(simulated_fv - analytical_fv)
relative_error = absolute_error / analytical_fv

print("ANALYTICAL VERIFICATION")
print("-----------------------")
print(f"Analytical future value: LKR {analytical_fv:,.2f}")
print(f"Simulated future value:  LKR {simulated_fv:,.2f}")
print(f"Absolute error: LKR {absolute_error:.10f}")
print(f"Relative error: {relative_error:.12%}")

# Verification assertion
assert np.isclose(
    simulated_fv,
    analytical_fv,
    rtol=1e-10,
    atol=1e-6
), "Verification failed!"

print("\nVerification PASSED!")
```

```python
# Verification 2: Zero investment return

zero_return_balance = 0.0

for month in range(MONTHS):
    zero_return_balance = (
        zero_return_balance * (1 + 0.0)
        + MONTHLY_CONTRIBUTION
    )

# Expected result: total contributions
expected_balance = MONTHLY_CONTRIBUTION * MONTHS

print("ZERO-RETURN VERIFICATION")
print("------------------------")
print(f"Expected savings: LKR {expected_balance:,.2f}")
print(f"Simulated savings: LKR {zero_return_balance:,.2f}")

assert np.isclose(
    zero_return_balance,
    expected_balance,
    rtol=0,
    atol=1e-6
), "Zero-return verification failed!"

print("\nZero-return verification PASSED!")
```

### Experiment 2: Sensitivity Analysis

Sensitivity analysis investigates how changes in key model parameters affect the probability of achieving the savings target.

Three parameters are examined:

1. Monthly savings contribution
2. Mean annual investment return
3. Mean annual inflation rate

Each parameter is varied independently while the other parameters remain at their baseline values.

The experiments use 10,000 simulation paths and a fixed random seed to support reproducibility.

The same underlying random shocks are reused across scenarios to reduce Monte Carlo noise when comparing parameter changes.

The results help identify which assumptions have the greatest influence on the probability of achieving the inflation-adjusted savings target.

```python
# Experiment 2: Sensitivity analysis

def calculate_success_probability(
    contribution=15000,
    annual_return=0.08,
    annual_inflation=0.05
):
    # Reuse the same random shocks for fair comparison
    return_shocks = (
        monthly_returns - monthly_return_mean
    ) / monthly_return_std

    inflation_shocks = (
        monthly_inflation - monthly_inflation_mean
    ) / monthly_inflation_std

    # Apply the selected scenario assumptions
    scenario_returns = (
        annual_return / 12
        + return_shocks * monthly_return_std
    )

    scenario_inflation = (
        annual_inflation / 12
        + inflation_shocks * monthly_inflation_std
    )

    # Start all scenarios with zero savings
    balances = np.zeros(N_SIMULATIONS)
    inflation_index = np.ones(N_SIMULATIONS)

    # Simulate monthly savings
    for month in range(MONTHS):
        balances = (
            balances * (1 + scenario_returns[:, month])
            + contribution
        )

        inflation_index *= (
            1 + scenario_inflation[:, month]
        )

    # Convert final nominal savings to real savings
    final_real = balances / inflation_index

    # Return probability of reaching the target
    return np.mean(final_real >= TARGET)


# Test different monthly contributions
contributions = [10000, 15000, 20000, 25000, 30000]

contribution_probs = [
    calculate_success_probability(contribution=c)
    for c in contributions
]

# Test different annual investment returns
annual_returns = [0.04, 0.06, 0.08, 0.10, 0.12]

return_probs = [
    calculate_success_probability(annual_return=r)
    for r in annual_returns
]

# Test different annual inflation rates
annual_inflations = [0.02, 0.04, 0.05, 0.06, 0.08]

inflation_probs = [
    calculate_success_probability(annual_inflation=i)
    for i in annual_inflations
]

# Display results
print("SENSITIVITY ANALYSIS RESULTS")
print("----------------------------")

print("\nMonthly Contribution:")
for c, p in zip(contributions, contribution_probs):
    print(f"LKR {c:,}: {p:.2%}")

print("\nAnnual Investment Return:")
for r, p in zip(annual_returns, return_probs):
    print(f"{r:.0%}: {p:.2%}")

print("\nAnnual Inflation:")
for i, p in zip(annual_inflations, inflation_probs):
    print(f"{i:.0%}: {p:.2%}")

# Check that the baseline scenario is reproduced
baseline_probability = calculate_success_probability()

assert np.isclose(
    baseline_probability,
    success_probability
), "Baseline sensitivity result does not match!"

print("\nBaseline consistency check PASSED!")
```

### Experiment 4: Continuous-Time Savings Model Using Euler's Method

A deterministic continuous-time savings model can be represented by the ordinary differential equation:

$$\frac{dS}{dt}=rS+c$$

Where:

- $S(t)$ = accumulated savings at time $t$
- $r$ = constant annual continuously compounded investment rate
- $c$ = annual contribution rate
- $t$ = time in years

The model assumes continuous contributions and constant returns.

Using Forward Euler:

$$S_{k+1}=S_k+\Delta t(rS_k+c)$$

For zero initial savings and nonzero $r$, the exact continuous-time solution is:

$$S(t)=\frac{c}{r}(e^{rt}-1)$$

For $r=0$, the exact solution is $S(t)=ct$.

The experiment compares Euler approximations at different time-step sizes against the exact solution.

This continuous-time model differs from the main monthly savings model, which uses discrete end-of-month contributions and monthly compounded returns.

```python
# Experiment 4: Euler method for continuous-time savings

annual_rate = 0.08
annual_contribution = MONTHLY_CONTRIBUTION * 12
duration_years = YEARS

def euler_savings(dt):
    steps = int(round(duration_years / dt))
    actual_dt = duration_years / steps

    savings = 0.0

    for _ in range(steps):
        savings += actual_dt * (
            annual_rate * savings + annual_contribution
        )

    return savings

# Exact continuous-time solution
exact_savings = (
    annual_contribution / annual_rate
    * (np.exp(annual_rate * duration_years) - 1)
)

time_steps = [1.0, 0.5, 0.25, 1/12, 1/24, 1/48]

euler_results = []
euler_errors = []

print("EULER METHOD COMPARISON")
print("-----------------------")
print(f"Exact solution: LKR {exact_savings:,.2f}\n")

for dt in time_steps:
    estimate = euler_savings(dt)
    error = abs(estimate - exact_savings)

    euler_results.append(estimate)
    euler_errors.append(error)

    print(
        f"dt = {dt:.5f} years | "
        f"Euler = LKR {estimate:,.2f} | "
        f"Error = LKR {error:,.2f}"
    )

# Verification: finer time steps reduce error
assert euler_errors[-1] < euler_errors[0]

print("\nEuler refinement check PASSED!")
```

```python
# Plot Euler approximation error against step size

plt.figure(figsize=(9, 5))

plt.loglog(
    time_steps,
    euler_errors,
    marker="o",
    linewidth=2
)

plt.xlabel("Time Step (Years)")
plt.ylabel("Absolute Error (LKR)")
plt.title("Forward Euler Error vs Time-Step Size")
plt.grid(True, which="both", alpha=0.3)
plt.tight_layout()
plt.show()
```

### Euler Method Comparison Interpretation

The Forward Euler experiment investigated the numerical approximation of a continuous-time savings accumulation model.

The exact continuous-time solution produced a final savings value of LKR 2,757,467.09 after 10 years.

Using an annual time step, Euler's method produced an absolute error of LKR 149,885.85. Reducing the time step to 1/12 year reduced the error to LKR 13,276.56, while a time step of 1/48 year reduced the error further to LKR 3,333.50.

The results demonstrate numerical convergence: smaller time steps produce approximations closer to the exact continuous-time solution.

The error decreases approximately in proportion to the time-step size, consistent with the first-order accuracy of Forward Euler's method.

The monthly Euler approximation also matches the deterministic monthly savings recurrence because both use the same update equation under the selected parameter conversion.

This comparison demonstrates the application of numerical ODE integration to savings modelling. The continuous-time experiment is supplementary to the main stochastic Monte Carlo model, which uses monthly contributions and uncertain returns.

### Discussion

This project investigated the probability of achieving an inflation-adjusted savings target using Monte Carlo simulation.

The baseline simulation estimated a 19.00% probability of accumulating LKR 2,000,000 in today's purchasing power after 10 years when saving LKR 15,000 per month.

The mean final real savings was LKR 1,673,494.37, while the median was LKR 1,618,263.52. These results indicate that the baseline contribution is unlikely to achieve the target under the assumed financial conditions.

Sensitivity analysis demonstrated that monthly contributions, investment returns, and inflation significantly influence the estimated probability of success.

Increasing monthly contributions to LKR 25,000 raised the estimated probability to 89.81%, while increasing the assumed annual investment return to 12% raised it to 52.44%. Increasing annual inflation to 8% reduced the probability to 1.98%.

The Monte Carlo convergence experiment showed that probability estimates became more stable as the number of simulations increased, consistent with the Law of Large Numbers.

Deterministic verification tests confirmed that the monthly savings recurrence matched the analytical future-value formula. The Forward Euler comparison additionally demonstrated numerical convergence when approximating a continuous-time savings model.

### Limitations

**1. Assumed Financial Distributions**

Investment returns and inflation were generated using normal distributions with illustrative parameter values. These assumptions have not been calibrated against historical Sri Lankan financial data.

**2. Independence Assumption**

The baseline model assumes that monthly investment returns and inflation shocks are independent. Real financial variables may exhibit correlation and temporal dependence.

**3. Constant Monthly Contributions**

The model assumes a fixed nominal contribution of LKR 15,000 per month. It does not account for salary growth, changing savings behaviour, or inflation-adjusted contributions.

**4. Excluded Financial Costs**

Investment fees, taxes, and transaction costs are excluded. Including these costs could reduce final savings.

**5. Simplified Investment Model**

The model assumes that savings earn returns according to a single stochastic process. It does not explicitly represent different asset classes, investment strategies, or changing market regimes.

**6. Monte Carlo Sampling Error**

The estimated success probability is based on 10,000 simulations and therefore contains sampling uncertainty. Increasing the simulation count can reduce this error but cannot eliminate uncertainty caused by model assumptions.

**7. Simplified Inflation Dynamics**

Inflation is modelled using independent monthly random values, rather than a more realistic process with persistent inflation trends or sudden economic shocks.

### Future Improvements

Potential extensions include:

- Calibrating investment returns and inflation using historical Sri Lankan financial data.
- Introducing correlation between investment returns and inflation.
- Modelling inflation-adjusted monthly contributions.
- Including investment fees and taxes.
- Comparing alternative investment strategies.
- Testing alternative probability distributions and economic scenarios.

### Conclusion

The simulation demonstrates how Monte Carlo methods can quantify uncertainty in long-term savings planning.

Under the illustrative baseline assumptions, the estimated probability of reaching the inflation-adjusted LKR 2,000,000 target is 19.00%.

The experiments show that higher monthly contributions and investment returns improve the probability of success, while higher inflation reduces it.

The results illustrate the usefulness of stochastic simulation for exploring financial planning decisions, while emphasizing the importance of model assumptions and uncertainty.
